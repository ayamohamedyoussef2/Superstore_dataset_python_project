#  **Data cleaning**

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df[["Sales", "Quantity", "Discount", "Profit"]].describe()
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
df["Shipping Day Delay"] = (df["Ship Date"] - df["Order Date"]).dt.days
```

```python
def get_profitability_status(profit):
    if profit > 0:
        return "Profitable"
    elif profit < 0:
        return "Loss"
```

```python
def get_discount_level(discount):
    if discount == 0:
        return "No Discount"
    elif discount < 0.30:
        return "Low Discount"
    elif discount >= 0.30:
        return "High Discount"
```

#  **Business questions**

What is the total sales generated by the business?

```python
total_sales = df["Sales"].sum()
print("Total Sales",total_sales)
```

How many orders were placed by customers?

```python
total_orders = df["Order ID"].nunique()
print("Total orders",total_orders)
```

What is the overall profit margin of the business?

```python
total_profit = df["Profit"].sum()
total_sales = df["Sales"].sum()
profit_margin = (total_profit / total_sales) * 100
print("profit margin",profit_margin)
```

How many products were sold in total?

```python
total_quantity = df["Quantity"].sum()
print("total quantity",total_quantity)
```

How do sales and profit vary across different product sub categories?

```python
subcategory_analysis = df.groupby("Sub-Category").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).reset_index()
```

How does profit vary across different shipment modes?

```python
shipping_analysis = df.groupby("Ship Mode").agg(
    Profit=("Profit", "sum")
).reset_index()
```

Which top 5 specific products generate the most overall revenue?

```python
t5 = df.groupby("Product Name").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).reset_index()
t5_top5 = t5.sort_values("Sales",ascending=False).head(5)
print(t5_top5)
```

Which months or quarters of the year consistently bring in the highest sales spike?

```python
month_q = df.groupby(["Year", "Month", "Month Name"]).agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).reset_index()
month_q = month_q.sort_values(["Year", "Month"])
print(month_q)
```

Which regions or states are making the least profit?

```python
region_q = df.groupby("Region").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")).reset_index()
region_q = region_q.sort_values("Profit")
print(region_q)

state_q = df.groupby("State").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).reset_index()
state_q = state_q.sort_values("Profit")
print("\n",state_q.head(10))
```

Which customer group spends the most per order?

```python
q7_orders = df.groupby(
    ["Segment", "Order ID"]).agg(Order_Sales=("Sales", "sum")).reset_index()
q7 = q7_orders.groupby("Segment").agg(
    Total_Sales=("Order_Sales", "sum"),Number_of_Orders=("Order ID", "nunique"),
    Average_Order_Value=("Order_Sales", "mean")
).reset_index()

print(q7)
```

Do bigger discounts actually bring in more sales volume?

```python
dis8 = df.groupby("Discount Level").agg(
    Sales=("Sales", "sum"),
    Quantity=("Quantity", "sum"),
    Orders=("Order ID", "nunique"),
    Profit=("Profit", "sum")).reset_index()

print(dis8)
```

Which categories and subcategories generate the most sales and profit?

```python
q1_category = df.groupby("Category").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")).reset_index()

print(q1_category)
```

```python
q1_subcategory = df.groupby("Sub-Category").agg(Sales=("Sales", "sum"),Profit=("Profit", "sum"),Quantity=("Quantity", "sum")).reset_index()
q1_subcategory = q1_subcategory.sort_values("Sales",ascending=False)
print(q1_subcategory)
```
